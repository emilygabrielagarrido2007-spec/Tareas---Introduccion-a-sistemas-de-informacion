## Introducción

En 1912, el personal encargado del embarque del Titanic utilizaba listas en papel para registrar a los pasajeros. En esta práctica, estudiaremos ese registro como un ejemplo de procesamiento manual de transacciones. Las listas cumplían una función operativa: dejar constancia de quiénes viajaban. El dataset reúne datos de los pasajeros, como su clase y puerto de embarque, e incorpora un dato conocido después del naufragio: si sobrevivieron.

### Diccionario de datos

El archivo `titanic.csv` contiene **891 registros y 12 columnas**. Cada fila representa a un pasajero.

| Columna | Tipo de dato | Descripción |

| `PassengerId` | Identificador entero | Código único asignado a cada pasajero en el dataset. |

| `Pclass` | Categoría ordinal | Clase del boleto. `1` = primera, `2` = segunda, `3` = tercera. |
| `Name` | Texto | Nombre del pasajero tal como aparece en el archivo. |
| `Sex` | Categoría | Sexo registrado. Los valores del archivo son `female` y `male`. |
| `Age` | Número | Edad en años. Puede incluir valores fraccionarios, por ejemplo, para bebés menores de un año. |
| `SibSp` | Conteo entero | Número de hermanos, hermanas y cónyuges del pasajero a bordo. |
| `Parch` | Conteo entero | Número de padres, madres, hijos e hijas del pasajero a bordo. |
| `Ticket` | Código de texto | Código del boleto. Puede incluir letras, números y signos, y repetirse en los registros de varios pasajeros. |
| `Fare` | Número decimal | Tarifa registrada en la fila del pasajero. |
| `Cabin` | Código de texto | Identificación de la cabina, cuando figura en el registro. |
| `Embarked` | Categoría | Puerto de embarque. `C` = Cherburgo, `Q` = Queenstown, `S` = Southampton. |

**3. ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?**

Analiza en la columna de "Embarked" y realiza el conteo de los embarques por classe, posteriormente calcula los vacios, e imprimimos la información en una tabla

```python
embarqueC=0
embarqueQ=0
embarqueS=0
noEmbarque=0

for embarque in df["Embarked"]:

    if embarque == "C":
        embarqueC+=1
    elif embarque == "Q":
        embarqueQ+=1
    elif embarque == "S":
        embarqueS+=1
    else:
        noEmbarque +=1


tabla_embarques = pd.DataFrame({
    "Embarque": ["Cherburgo", "Queenstown", "Southampton", "Sin Registrar"],
    "Cantidad_Pasajeros": [embarqueC, embarqueQ, embarqueS, noEmbarque]
})

display(tabla_embarques)
```

**4. ¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?**

En cada lista se van guardando todas las tarifas segun la clase. Luego se procede a realizar los cálculos solicitados, y se crea una tabla para mostrar la información de las tarifas por clase

```python
tarifa1=[]
tarifa2=[]
tarifa3=[]

for clases, tarifa in zip(df["Pclass"], df["Fare"]):
    if clases == 1:
        tarifa1.append(tarifa)

    elif clases == 2:
        tarifa2.append(tarifa)
    else:
        tarifa3.append(tarifa)


promedio1 = sum(tarifa1)/len(tarifa1)
maxim1 = max(tarifa1)
minim1 = min(tarifa1)

promedio2 = sum(tarifa2)/len(tarifa2)
maxim2 = max(tarifa2)
minim2 = min(tarifa2)

promedio3 = sum(tarifa3)/len(tarifa3)
maxim3 = max(tarifa3)
minim3 = min(tarifa3)


tablaResTarifa = pd.DataFrame({
    "Métricas": ["Promedio", "TarifaMaxima", "TarifaMinima"],
    "Clase 1": [promedio1, maxim1, minim1],
    "Clase 2": [promedio2, maxim2, minim2],
    "Clase 3": [promedio3, maxim3, minim3],
})

display(tablaResTarifa)
```

**5. ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?**

Se crearon arreglos para cada embarcación, en los cuales se van añadiendo números segun la clase que haya embarcado ahí. Al finalizar el recorrido de todos los elementos, se crea la tabla en la que directamente, se van contando cuantos de primera, segunda y tercera clase embarcaron en su puerto respectivo y que clase también es la que no embarcó.

```python
contEmbC=[]
contEmbQ=[]
contEmbS=[]
contSinEmb=[]


for embarque, clases in zip(df["Embarked"], df["Pclass"]):
    if embarque == "C":
        contEmbC.append(clases)
    elif embarque == "Q":
        contEmbQ.append(clases)
    elif embarque == "S":
        contEmbS.append(clases)
    else:
        contSinEmb.append(clases)

tablaClasePuerto = pd.DataFrame({
    "Clase" : ["Clase 1", "Clase 2", "Clase 3"],
    "Cherburgo": [contEmbC.count(1), contEmbC.count(2), contEmbC.count(3)],
    "Queenstown": [contEmbQ.count(1), contEmbQ.count(2), contEmbQ.count(3)],
    "Southampton": [contEmbS.count(1), contEmbS.count(2), contEmbS.count(3)],
    "Sin Registrar": [contSinEmb.count(1), contSinEmb.count(2), contSinEmb.count(3)]
})

display (tablaClasePuerto)



```

**6. ¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?**

Al igual que en el caso anterior, se analizan a la par el estado de supervivencia y la clase, cada recorrido va añadiendo la clase que sobrevivió o falleció a cada estado/lista. Al finalizar, se crea una tabla en la que se cuenta directamente cuantas personas de cada clase están en cada categoría

```python
sobrevivieron = []
fallecieron = []


for survive, clases in zip(df["Survived"], df["Pclass"]):
    if survive:
        sobrevivieron.append(clases)
    else:
        fallecieron.append(clases)


tablaContSupervivientes = pd.DataFrame({
    "Estado": ["Clase 1", "Clase 2", "Clase 3"],
    "Sobrevivieron": [sobrevivieron.count(1), sobrevivieron.count(2), sobrevivieron.count(3)],
    "Fallecieron":[fallecieron.count(1), fallecieron.count(2), fallecieron.count(3)]
})

display(tablaContSupervivientes)
```

**7. ¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?**

Sumé las columnas con la información de los acompañantes de los pasajeros para poder realizar en un solo bucle los cálculos, de ahí se usaron los contadores para poder tabular cada tipo de pasajero con sus familiares acompañados y mostrar la tabla realizada

```python
totalFamiliares = df["SibSp"] + df["Parch"]

contPasajerosSolos=0
contPasajerosUnFamiliar=0
contPasajerosDosoMas=0


for parch in totalFamiliares:
    if parch == 0:
        contPasajerosSolos+=1
    elif parch == 1:
        contPasajerosUnFamiliar+=1
    else:
        contPasajerosDosoMas+=1


tablaPasajFamiliares = pd.DataFrame({
    "Familiares": ["Cero", "Uno", "Dos o Mas"],
    "Pasajeros": [contPasajerosSolos, contPasajerosUnFamiliar, contPasajerosDosoMas]
})

display(tablaPasajFamiliares)
```

**8. ¿Cuántos pasajeros tenían menos de 18 años, cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.**

Realicé 3 listas, una para menores de 18 años, otra para los que tenían 18 o más y para los que no tenían la edad registrada. En cada iteración del for, se va comparando que clases entran en cada categoría y se añaden al arreglo. Al final se realiza el conteo por categoría y por clase y se lo tabuló, para luego ser mostrado.

```python
menor18=[]
mayor18=[]
sinRegistrar=[]

for clases, edad in zip (df["Pclass"], df["Age"]):
    if edad < 18:
        menor18.append(clases)

    elif edad > 18 or edad == 18:
        mayor18.append(clases)
    else:
        sinRegistrar.append(clases)


tablaEdades=pd.DataFrame({
    "Edades": ["Clase 1", "Clase 2", "Clase 3"],
    "Menores a 18": [menor18.count(1), menor18.count(2), menor18.count(3)],
    "18 años o más": [mayor18.count(1), mayor18.count(2), mayor18.count(3)],
    "Edad sin registrar": [sinRegistrar.count(1), sinRegistrar.count(2), sinRegistrar.count(3)]
})

display(tablaEdades)
```

**9. ¿Cuántos códigos de boleto distintos hay? ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?**

Creo una especie de tabla con el número de veces que se repite cada boleto, cuento cuantos sin repetir, y al final voy iterando para revisar cuántos están asociados a más de un pasajero y cual de ellos es el que aparece con mayor frecuencia.  

```python
conteoBoletos = df["Ticket"].value_counts()

cantidadBoletosDis = len(conteoBoletos)

boletosAsociados=0
masRepetido=0
boletoFrecuente=""

for boleto in conteoBoletos.index:
    cantidad = conteoBoletos[boleto] #para ver la frecuencia

    if cantidad > 1 :
        boletosAsociados +=1 

    if cantidad > masRepetido:
        masRepetido = cantidad
        boletoFrecuente= boleto



tablaBoletos = pd.DataFrame({
    "Codigos Boleto Distinto": [cantidadBoletosDis],
    "Asociados a más de un pasajero": [boletosAsociados],
    "Boleto en más registros": [boletoFrecuente]
})

display(tablaBoletos)
```

**10. ¿En qué clase hay más registros sin cabina? Para cada clase, muestren cuántos registros tienen cabina, cuántos carecen de ella y cuántos hay en total.**

Creé dos arreglos para cada estado, utilicé pd.isna para saber cuales eran las que no tenían cabina, y fui armando la tabla realizando el conteo de cada estado por clase. Al final comparé todos los que no tenían registros e imprimí laa clase con mayor registros sin cabina.

```python
conCabina=[]
sinCabina=[]

for clase, cabina in zip(df["Pclass"], df["Cabin"]):
    if pd.isna(cabina):
        sinCabina.append(clase)
    else:
        conCabina.append(clase)

tablaCabinas=pd.DataFrame({
    "Clase": ["Clase 1", "Clase 2", "Clase 3"],
    "Con Cabina": [conCabina.count(1), conCabina.count(2), conCabina.count(3) ],
    "Sin Cabina": [sinCabina.count(1), sinCabina.count(2), sinCabina.count(3) ],
    "Total": [conCabina.count(1)+sinCabina.count(1), conCabina.count(2)+sinCabina.count(2), conCabina.count(3)+sinCabina.count(3)]
})
display(tablaCabinas)

if sinCabina.count(3) > sinCabina.count(1) and sinCabina.count(3) > sinCabina.count(2):
    print("La clase con más registros sin cabina es la Clase 3.")
elif sinCabina.count(2) > sinCabina.count(1) and sinCabina.count(2) > sinCabina.count(3):
    print("La clase con más registros sin cabina es la Clase 2.")
else:
    print("La clase con más registros sin cabina es la Clase 1.")
```
